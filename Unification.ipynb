```python
def is_variable(x):
    return isinstance(x, str) and x.islower()
def unify(x, y, substitution):
    
    if x == y:
        return substitution
   
    if is_variable(x):
        substitution[x] = y
        return substitution

    if is_variable(y):
        substitution[y] = x
        return substitution
 
    return None
def unification(expr1, expr2):
  
    if expr1[0] != expr2[0]:
        return "FAIL"

    if len(expr1[1]) != len(expr2[1]):
        return "FAIL"
    substitution = {}
   
    for x, y in zip(expr1[1], expr2[1]):
        result = unify(x, y, substitution)
        if result is None:
            return "FAIL"
    return substitution
E1 = ("Eats", ["x", "Apple"])
E2 = ("Eats", ["Riya", "y"])

result = unification(E1, E2)

if result == "FAIL":
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)
    print("Unified Expression: Eats(Riya, Apple)")
```
