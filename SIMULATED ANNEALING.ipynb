```python
import random
import math

N = 8

def calculate_cost(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing():
    while True:
        board = [random.randint(0, N - 1) for _ in range(N)]

        temperature = 100.0
        cooling_rate = 0.99

        while temperature > 0.01:
            current_cost = calculate_cost(board)

            if current_cost == 0:
                 return board

            new_board = board.copy()
            col = random.randint(0, N - 1)
            new_row = random.randint(0, N - 1)

            new_board[col] = new_row
            new_cost = calculate_cost(new_board)

            if new_cost < current_cost:
                board = new_board

            elif random.random() < math.exp((current_cost - new_cost) / temperature):
                board = new_board

            temperature *= cooling_rate



solution = simulated_annealing()

print("Solution found")
print("Queen positions :", solution)
print("Number of conflicts:", calculate_cost(solution))

print("\nChessboard:")

for row in range(N):
    for col in range(N):
        if solution[col] == row:
            print("Q", end=" ")
        else:
            print(".", end=" ")
    print()
```
