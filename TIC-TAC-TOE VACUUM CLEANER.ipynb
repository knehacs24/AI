```python
import random

def print_board(board):
    print("\n")
    for row in board:
        print(" | ".join(row))
        print("-" * 9)
    print("\n")

def check_win(board, player):
    for i in range(3):
        if all(board[i][j] == player for j in range(3)) or \
           all(board[j][i] == player for j in range(3)):
            return True
    if all(board[i][i] == player for i in range(3)) or \
       all(board[i][2 - i] == player for i in range(3)):
        return True
    return False

def check_tie(board):
    return all(cell != " " for row in board for cell in row)

def computer_move(board):
    empty_cells = [(r, c) for r in range(3) for c in range(3) if board[r][c] == " "]
    if empty_cells:
        return random.choice(empty_cells)
    return None

def play_game():
    board = [[" " for _ in range(3)] for _ in range(3)]
    turn = "X"  
   
    while True:
        print_board(board)
       
        if turn == "X":
            print("Your turn (X)")
            try:
                row = int(input("Enter row (0, 1, or 2): "))
                col = int(input("Enter col (0, 1, or 2): "))
                if board[row][col] != " ":
                    print("Spot taken! Choose another.")
                    continue
            except (ValueError, IndexError):
                print("Invalid input. Enter numbers between 0 and 2.")
                continue
           
            board[row][col] = "X"
            if check_win(board, "X"):
                print_board(board)
                print("You win!")
                break
            turn = "O"
        else:
            print("Computer's turn (O)...")
            move = computer_move(board)
            if move:
                board[move[0]][move[1]] = "O"
            if check_win(board, "O"):
                print_board(board)
                print("Computer wins!")
                break
            turn = "X"
           
        if check_tie(board):
            print_board(board)
            print("It's a tie!")
            break

if __name__ == "__main__":
    play_game()
```

```python
import time
import random

class VacuumEnvironment:
    def __init__(self):
        self.locations = {'A': 'Clean', 'B': 'Clean'}
        self.agent_location = random.choice(['A', 'B'])
        self.randomize_dirt()

    def randomize_dirt(self):
        for room in self.locations:
            self.locations[room] = random.choice(['Clean', 'Dirty'])

    def get_percept(self):
        return self.agent_location, self.locations[self.agent_location]

    def execute_action(self, action):
        print(f"[Robot Action]: {action}")
        if action == 'Suck':
            self.locations[self.agent_location] = 'Clean'
        elif action == 'MoveRight':
            self.agent_location = 'B'
        elif action == 'MoveLeft':
            self.agent_location = 'A'

    def display_status(self):
        print(f"Room A: {self.locations['A']} | Room B: {self.locations['B']} | Robot is in Room: {self.agent_location}")
        print("-" * 60)

class ReflexVacuumAgent:
    def program(self, percept):
        location, status = percept

        if status == 'Dirty':
            return 'Suck'
        elif location == 'A':
            return 'MoveRight'
        elif location == 'B':
            return 'MoveLeft'

if __name__ == "__main__":
    environment = VacuumEnvironment()
    agent = ReflexVacuumAgent()
    print("Initial Environment State:")
    environment.display_status()


    for step in range(1, 6):
        print(f"--- Cycle Step {step} ---")
       

        percept = environment.get_percept()
        print(f"Perceived: Location {percept[0]} is {percept[1]}")

        action = agent.program(percept)

        environment.execute_action(action)

        environment.display_status()
        time.sleep(1)
```
